```python
import tensorflow
from tensorflow.keras.models import load_model
import pickle
import joblib
import numpy as np
import pandas as pd
```

```python
### Train the model , scaler pikle,onehot 
model=load_model('model.h5')  # Load the trained model from a file

#load the encoder and scaler
with open('one_hot_encoder.pkl', 'rb') as file:
        one_hot_encoder = pickle.load(file)  # Load the one-hot encoder from a pickle file

with open('label_encoder.pkl', 'rb') as file:
    label_encoder = pickle.load(file)  # Load the label encoder from a pickle file
    
with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)  # Load the scaler from a pickle file
```

```python
# Load trained ANN model
model = load_model('model.h5')

# Load OneHotEncoder
one_hot_encoder = joblib.load('one_hot_encoder.pkl')

# Load LabelEncoder
label_encoder = joblib.load('label_encoder.pkl')

# Load StandardScaler
scaler = joblib.load('scaler.pkl')

print("All files loaded successfully!")
```

```python
## Exapmple input data for prediction
input_data = {
    'CreditScore':700,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000}
```

```python
## One Hot encoder the 'Geography column
geography_encoded = one_hot_encoder.transform([[input_data['Geography']]]).toarray()
geography_encoded_df = pd.DataFrame(geography_encoded, columns=one_hot_encoder.get_feature_names_out(['Geography']))

geography_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
## Concatenate one hot encoded
input_df=pd.concat([input_df.drop('Geography', axis=1), geography_encoded_df], axis=1)

input_df
```

```python
## Encode Categorical Vaariables
input_df['Gender'] = label_encoder.transform(input_df['Gender'])

input_df
```

```python
## Scaling the inout the data
input_sc = scaler.transform(input_df)
input_sc
```

```python
# Prediction churn Probability
Churn_prediction = model.predict(input_sc)
Churn_prediction 
```

```python
churn_proba=Churn_prediction[0][0]
churn_proba
```

```python
if churn_proba > 0.5:
    print("The customer is likely to churn.")
else:
    print("The customer is not likely to churn.")
```
